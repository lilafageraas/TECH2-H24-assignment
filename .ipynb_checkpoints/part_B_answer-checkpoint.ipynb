```python
import csv

#open csv file (data from boss) in reading mode
with open('data.csv', mode='r') as file:
    #open csv file in reading mode
    csv_reader = csv.reader(file)
    #convert the csv reader to a list of rows
    data = list(csv_reader)

#make each column a separate list
column1 = [float(row[0]) for row in data if row[0].strip() != '']
column2 = [float(row[1]) for row in data if row[1].strip() != '']
column3 = [float(row[2]) for row in data if row[2].strip() != '']
```

```python
#standard deviation method 1
from part_A import std_loops

print(f'the standard deviation of list 1 using method 1 is {std_loops(column1):.2f}')
print(f'the standard deviation of list 2 using method 1 is {std_loops(column2):.2f}')
print(f'the standard deviation of list 3 using method 1 is {std_loops(column3):.2f}')
```

```python
#standard deviation method 2
from part_A import std_builtin

print(f'the standard deviation of list 1 using method 2 is {std_builtin(column1):.2f}')
print(f'the standard deviation of list 2 using method 2 is {std_builtin(column2):.2f}')
print(f'the standard deviation of list 3 using method 2 is {std_builtin(column3):.2f}')
```

```python
#standard deviation method 3
import numpy as np

print(f'the standard deviation of list 1 using method 3 is {np.std(column1):.2f}')
print(f'the standard deviation of list 2 using method 3 is {np.std(column2):.2f}')
print(f'the standard deviation of list 3 using method 3 is {np.std(column3):.2f}')
```

```python
# recording the run time for method 1
%timeit std_loops(column1)
%timeit std_loops(column2)
%timeit std_loops(column3)
```

```python
# recording the run time for method 2
%timeit std_builtin(column1)
%timeit std_builtin(column2)
%timeit std_builtin(column3)
```

```python
# recording the run time for method 3
%timeit np.std(column1)
%timeit np.std(column2)
%timeit np.std(column3)
```

# summary:
for little amounts of data (sequence <100 numbers), it is best to use method 2 (std_builtin)
for 100 < data < 1000, it is best to use method 3 (np.std)
for 1000 < data < 10 000, it is best to use method 3 (np.std)

the fastest approach depends on the amount of data, this might be because the methods are optimixed for different things
